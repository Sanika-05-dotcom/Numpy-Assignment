```python








import numpy as np



months = np.arange(1, 13)
stores = np.arange(1, 11)


sales = np.random.randint(5000, 20001, size=(10, 12))

print("MONTHLY SALES DATA")
print(sales)


Q1 = sales[:, 0:3]
Q2 = sales[:, 3:6]
Q3 = sales[:, 6:9]
Q4 = sales[:, 9:12]

print("\nQ1 Sales", Q1)
print("\nQ2 Sales", Q2)
print("\nQ3 Sales", Q3)
print("\nQ4 Sales", Q4)

print("\nStore 1 Monthly Sales:", sales[0, :])


target = 12000

print("\nSales Greater Than Target:")
print(sales[sales > target])


store_avg = np.mean(sales, axis=1)
overall_avg = np.mean(sales)

print("\nStores Below Overall Average:")
print(stores[store_avg < overall_avg])


high_sales = (sales > 15000) & (sales < 19000)
print("\nSales Between 15000 and 19000:")
print(sales[high_sales])


selected_sales = (sales < 7000) | (sales > 18000)
print("\nSales Below 7000 OR Above 18000:")
print(sales[selected_sales])


not_high_sales = ~(sales > 18000)
print("\nSales NOT Greater Than 18000:")
print(sales[not_high_sales])


selected_stores = sales[[1, 4, 7], :]
print("\nSales of Stores 2, 5, and 8:")
print(selected_stores)



monthly_total = np.sum(sales, axis=0)
best_month_indices = np.argsort(monthly_total)[-3:][::-1]

print("\nTop 3 Months (Month Numbers):")
print(months[best_month_indices])

print("\nSales Totals of Top 3 Months:")
print(monthly_total[best_month_indices])


growth_targets = np.linspace(10000, 15000, 12)

print("\nMonthly Growth Targets:")
print(np.round(growth_targets, 2))

print("\nMONTHLY SALES REPORT")
for i in range(12):
    status = "Target Met" if monthly_total[i] / 10 >= growth_targets[i] else "Target Not Met"
    print(
        f"Month {months[i]}: Total Sales = {monthly_total[i]}, "
        f"Average per Store = {monthly_total[i]/10:.2f}, "
        f"Target = {growth_targets[i]:.2f}, Status = {status}"
    )

# Store-wise annual report
annual_sales = np.sum(sales, axis=1)

print("\nSTORE-WISE ANNUAL REPORT")
for i in range(10):
    print(
        f"Store {stores[i]}: Annual Sales = {annual_sales[i]}, "
        f"Average Monthly Sales = {store_avg[i]:.2f}"
    )
```
